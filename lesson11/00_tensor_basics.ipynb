### Creating a tensor

### Convert from array to tensor

```python
x = np.array([1, 2, 3, 4, 5])

tensor_arr = torch.from_numpy(x)
print(tensor_arr)
```

```python
x = torch.tensor([1, 2, 3, 4, 5])

arr = x.numpy()

print(arr)
```

```python
x = torch.rand(3, 3) # generates random numbers between 0, 1
print(x)
```

```python
x = torch.ones(3, 3)
print(x)
```

```python
x = torch.zeros(3, 3)
print(x)
```

```python
print(x.size())
print(x.dtype)
print(x.ndim)
```

```python
x = torch.zeros(3, 3, dtype=torch.float16)
print(x.dtype)
```

### Tensor Operations

```python
x = torch.rand(3, 3)
y = torch.rand(3, 3)

print(x)
print(y)
print("*" * 50)
print(x + y) # element-wise addition

print("*" * 50)
print(x - y) # element-wise subtraction

print("*" * 50)
print(x * y) # element-wise multiply

print("*" * 50)
print(x / y) # element-wise division
```

```python
z = torch.add(x, y) # x + y
z = torch.sub(x, y) # x - y
z = torch.mul(x, y) # x * y
z = torch.div(x, y) # x / y

print(z)
```

### Tensor slicing

```python
x = torch.rand(5, 3)

print(x)
```

```python
# all rows, column 0
print(x[:, 0])

# row 1, all columns
print(x[1, :])

# element at 1,1
print(x[1, 1])
```

```python
print(x[1, 1].item())
```

### Reshape of Tensors

```python
x = torch.randn(4, 4)
print(x)

print("*" * 100)
y = x.view(16) # flatten the tensor
print(y)
```

```python
z = x.view(2, 8)
print(z)
```

```python
z = x.view(16, 1)
print(z)
```

```python
z = x.view(-1, 8)
print(z)
```

```python
z = x.view(8, -1)
print(z)
```

```python
z = x.view(-1) # -1 infers the other dimension as we did in numpy arrays
print(z)
```
