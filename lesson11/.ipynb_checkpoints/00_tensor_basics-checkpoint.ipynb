### Moving tensors from cpu to gpu and vice-versa

```python
x = torch.tensor([1.0, 2.0, 3.0])
print(x.device) # cpu


device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
# move the tensor to Mac GPU / mps
x_gpu = x.to(device)
print(x_gpu.device) # mps
```

```python
x = torch.tensor([1.0, 2.0, 3.0], device="cuda") # linux/windows
x = torch.tensor([1.0, 2.0, 3.0], device="mps") # MacOS

 # MacOS
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
x = torch.tensor([1.0, 2.0, 3.0], device=device)

# linux/windows
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
x = torch.tensor([1.0, 2.0, 3.0], device=device) 
```

### Move tensor from GPU to CPU

```python
x = torch.tensor([1.0, 2.0, 3.0])
print(x.device) # cpu


device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
# move the tensor to Mac GPU / mps
x_gpu = x.to(device)
print(x_gpu.device) # mps

x_cpu = x_gpu.to("cpu")
# x_cpu = x_gpu.cpu()
print(x_cpu.device) # cpu
```

### Complete Example

```python
# device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
print(device)

# create a tensor on CPU
X = torch.randn(3, 3)
print(X)

# move from CPU -> GPU
X_gpu = X.to(device)
print(X_gpu.device)

# perform computation on GPU
y_gpu = X_gpu * 2
print(y_gpu)

# GPU -> CPU
y_cpu = y_gpu.cpu()

print(f"Original: {X.device}")
print(f"CPU tensor: {y_cpu.device}")
```

### Important: CPU and GPU tensors cannot be mixed

```python
x_cpu = torch.randn(3, 3).to("cpu") # cpu
x_gpu = torch.randn(3, 3).to("mps") # mps

# RuntimeError: tensors are on different devices
result = x_cpu + x_gpu
print(result)
```

### Moving a neural network model
- move both the model and its input tensors to the same device
- device='mps:0' means you only have one GPU on your machine
- device='mps:1' means you have two GPU installed on your machine

```python
import torch
import torch.nn as nn

device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")

# Define model
model = nn.Sequential(
    nn.Linear(10, 64),
    nn.ReLU(),
    nn.Linear(64, 1)
)

print(model) # lives on CPU

# move model to GPU
model = model.to(device)

# create input and move it to GPU
X = torch.randn(32, 10).to(device)

# forward pass
output = model(X)
print(output)
```

```python
print(output.to("cpu"))
```

### Automatic Differentiation (AutoGrad)
- Pytorch automatically calculates gradients used during neural network training
- `y=x^2`
- `dy/dx = 2x`

```python
x = torch.tensor(3.0, requires_grad=True)
print(x)

# mathematical function
y = x ** 2

# calculate gradients
y.backward() # dy/dx = 2x

print(f"value: {y.item()}")
print(f"gradient: {x.grad}")
print(f"gradient: {x.grad.item()}")
```

```python
x = torch.tensor(3.0, requires_grad=True)
print(x)

# mathematical function
y = x ** 2 + 2*x + 1

# calculate gradients
y.backward() # dy/dx = 2x + 2

# when x = 3, dy/dx = 8

print(f"value: {y.item()}")
print(f"gradient: {x.grad.item()}")
```

### Mini Neural Network
- y = Ax+b
- `A` -> data points (Matrix)
- `b` -> called bias (column vector)
- `y = (10, 2)x + (10,1)`
- `bmi = (w_1*height + w_2 * weight) + b`
- `prediction_bmi = 2.2 * height + 3.5 * weight `
- `w_i` -> parameters/weights/coefficient

```python
weight = torch.tensor([80.0]) # kg
height = torch.tensor([1.75]) # meters

# bmi calculation
# bmi = weight / (height ** 2)
bmi = weight / (height.pow(2))

print(f"predicted bmi: {bmi.item()}")
```

```python
import numpy as np

weight = np.array([80.0]) # kg
height = np.array([1.75]) # meters

# bmi calculation
bmi = weight / (height ** 2)

print(f"predicted bmi: {bmi}")
```

### Mini Neural Network Calculation

```python
X = torch.tensor([2.0, 3.0, 4.0])

# weight matrix: 3 inputs, 2 neurons
W = torch.tensor([
    [0.2, 0.5],
    [0.3, 0.7],
    [0.4, 0.1]
])

print(W.shape)

# bias
b = torch.tensor([0.1, 0.2])
print(b.shape)

# weight = height * W + b
# linear function: z = XW + b, z = WX + b
# z = X @ W + b
z = W.T @ X + b
print(z)

# apply relu activation function
output = torch.relu(z) # max(0, x) -> max(0, 3) -> 3
print(output)
```

```python
output = torch.relu(torch.tensor(-3))
print(output)
```

```python
x = torch.tensor([-10, -5, -2, 0, 2, 5, 10])

# apply sigmoid
output = torch.sigmoid(x)
print(output)
```
