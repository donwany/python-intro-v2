### Mathematical equation
- $y = x^2$
- $dy/dx = 2x$

```python
x = torch.tensor(3.0, requires_grad=True)
print(x)

# mathematical function
y = x ** 2

# calculate gradients
y.backward() # dy/dx = 2x

print(f"value of y at x=3: {y.item()}")
print(f"gradient: {x.grad}")
```

### Linear Transformations
- $Y = XW^T + b$
- W -> weights, b -> bias
- Example: X -> [years_of_experience, number_of_certifications]
- X = in_features = 2 (each sample has 2 input features)
- Y = out_features = 3 (the layer produces 3 output values for each sample)

```python
# Input
import torch
import torch.nn as nn

layer = nn.Linear(2, 2)

print(layer)
print("*" * 50)
print(layer.weight.data) # W^T
print("*" * 50)
print(layer.bias.data) # b
print("*" * 50)

X = torch.tensor([
    [5.0, 3.0],
    [2.0, 1.0],
    [10.0, 4.0]
])

print(X)

# weight matrix
layer.weight.data = torch.tensor([
    [2.0, 1.0],
    [3.0, 4.0]
])

# bias
layer.bias.data = torch.tensor([10.0, 20.0])

Y = layer(X) # $Y = XW^T + b$

print("*" * 50)
print(Y)
```

```python
# suppose you have 100 students each represented by 2 features
# 100 = number of studnets
# 2 = input features per student
# 3 = output features per student

X = torch.randn(100, 2) # input has 2 columns

#print(X)

layer = nn.Linear(in_features=2, out_features=3, bias=True)

print(layer.weight.data) # 3x2
print(layer.bias.data)   # 1x3


print("*" * 50)
Y = layer(X)
print(Y) # 100x3 -> ouput has 3 columns
```
