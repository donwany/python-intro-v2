## Create a dataframe from dictionary

```python
data = {
    "Name": ["John", "Mary", "David"],
    "Age": [22, 25, 30],
    "temperature": [98.6, 99.1, 97.5],
    "City": ["Accra", "Houston", "Toronto"], 
    "Location": ["Ghana", "USA", "Canada"],
    "Occupation": ["Engineer", "Doctor", "Artist"]
}

df = pd.DataFrame(data)

df
```

## Filtering data

```python
df
```

```python
df['City']
```

```python
# those living in Houston
df[ df['City'] == 'Houston' ]
```

```python
# those not living in Houston
df[ df['City'] != 'Houston' ]
```

```python
# | means OR, & means AND

df[ (df['City'] == 'Accra') | (df['Age'] <= 25) ]
```

```python
df[ (df['City'] == 'Accra') & (df['Age'] <= 25) ]
```
