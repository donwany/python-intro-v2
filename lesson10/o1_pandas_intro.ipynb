### Student Data

```python
students = {
    "Name": ["John", "Mary", "David", "Sarah"],
    "Math": [80, 95, 70, 88],
    "Science": [85, 90, 75, 92]
}

df = pd.DataFrame(students)

df
```

```python
df['average_score'] = (df['Math'] + df['Science']) / 2

# find those who failed or passed with an average score of 75%
df['Passed'] = df['average_score'] >= 75

df
```

### Create a function and apply it to the average column

```python
# using this function approach

def determine_pass_fail(avg_score):
    if avg_score >= 75:
        return "Passed"
    else:
        return "Failed"


# using lambda approach (HOME WORK)

df['Status'] = df['average_score'].apply(determine_pass_fail)
```

```python
df
```

```python
df.to_csv("avg_score.csv", index=True)
```

### Merging DataFrames

```python
students = pd.DataFrame({
    "StudentID": [1, 2, 3],
    "Name": ["John", "Mary", "David"]
})

students
```

```python
scores = pd.DataFrame({
    "StudentID": [1, 2, 4],
    "Score": [85, 90, 75]
})

scores
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    "customer_id": np.random.choice(
        ["C001", "C002", "C003", "C004", "C005"],
        20
    ),
    "region": np.random.choice(
        ["North", "South", "East", "West"],
        20
    ),
    "product": np.random.choice(
        ["Laptop", "Phone", "Tablet", "Monitor"],
        20
    ),
    "quantity": np.random.randint(1, 6, 20),
    "price": np.random.choice(
        [299, 499, 799, 999, 1299],
        20
    ),
    "date": pd.date_range(
        "2026-01-01",
        periods=20,
        freq="D"
    )
})

df
```

```python
df['revenue'] = df['quantity'] * df['price']
df
```

```python
df[
    (df['quantity'] > 2) & (df['revenue'] > 1000)
]
```

```python
df.query("quantity > 2 and revenue > 1000")
```

```python
df.query("quantity >= 2 and price >=500 and region == 'North' ")
```

```python
df[
    (df['quantity'] >= 2) & (df['price'] >= 500) & (df['region'].isin(['North', 'West']))
]
```

### Method chaining
- Homework to run

```python
result = (
    df
    .query("quantity >= 2")
    .assign(revenue=lambda x: x["quantity"] * x["price"])
    .groupby("region")
    .agg(total_revenue=("revenue", "sum"), average_revenue=("revenue", "mean"))
    .sort_values("total_revenue",ascending=False)
    .reset_index()
)

result
```

```python
customer_summary = (
    df.groupby("customer_id")
      .agg(
          total_revenue=("revenue", "sum"),
          average_order=("revenue", "mean"),
          orders=("revenue", "count"),
          last_purchase=("date", "max")
      )
      .reset_index()
)

customer_summary["revenue_pct"] = (
    customer_summary["total_revenue"] / customer_summary["total_revenue"].sum() * 100
)

top10 = (customer_summary.nlargest(10, "total_revenue"))

top10
```
