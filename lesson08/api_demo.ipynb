```python
import requests
from pprint import pprint

def fetch_posts(page=1, limit=5):
    url="https://jsonplaceholder.typicode.com/posts"
    params = {
        "page": page,
        "limit": limit
    }
    response = requests.get(url, params=params, timeout=10)

    print(response.status_code)

    return response.json()


posts = fetch_posts(page=2)
pprint(posts)
```

```python
# celius to fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 1.8) + 32


def fahrenheit_to_celsius(fahrenheit_temp):
  celsius_temp = (fahrenheit_temp - 32) * 5/9
  return celsius_temp
```

```python
class NewsAPI:
    def __init__(self, api_key):
        self.api_key = api_key
        self.base_url = "https://newsapi.org/v2/everything"

    def fetch_news(self, query, from_date, sort_by="publishedAt"):
        url = f"{self.base_url}?q={query}&from={from_date}&sortBy={sort_by}&apiKey={self.api_key}"
        response = requests.get(url)
        return response.json()
```

```python
class NewsHeadlines:
    def __init__(self, api_key):
        self.api_key = api_key
        self.base_url = "https://newsapi.org/v2/top-headlines"

    def fetch_headlines(self, country="us", category="business"):
        url = f"{self.base_url}?country={country}&category={category}&apiKey={self.api_key}"
        print(f"Fetching headlines from URL: {url}")
        response = requests.get(url)
        return response.json()
```
